```python
import numpy as np

products = np.array([
    "Laptop", "Mobile", "Tablet",
    "Monitor", "Keyboard", "Headphones"
])

months = np.array(["Jan", "Feb", "Mar", "Apr"])

# Units sold: rows = products, columns = months
units_sold = np.array([
    [45, 52, 48, 60],
    [120, 135, 128, 150],
    [65, 58, 72, 80],
    [35, 40, 38, 32],
    [90, 85, 100, 115],
    [75, 95, 110, 105]
])

# Selling price per unit (₹)
selling_price = np.array([
    50000, 18000, 22000, 12000, 1500, 2500
])

# Cost price per unit (₹)
cost_price = np.array([
    42000, 15000, 19000, 9500, 900, 1700
])
```

```python
                                    Part A — Revenue & Profit Engine
                                            Core calculations
```

```python
# Q1. Har product ka total units sold nikalo (4 months combined).

print(f"Total Sold Unit of Laptop : {np.sum(units_sold[0])}")
print(f"Total Sold Unit of Mobile : {np.sum(units_sold[1])}")
print(f"Total Sold Unit of Tablet : {np.sum(units_sold[2])}")
print(f"Total Sold Unit of Monitor : {np.sum(units_sold[3])}")
print(f"Total Sold Unit of Keyboard : {np.sum(units_sold[4])}")
print(f"Total Sold Unit of Headphones : {np.sum(units_sold[5])}")
```

```python
# Q2. Har product ka total revenue nikalo.
product_units = np.sum(units_sold,axis = 1)
# for u in units_sold:
#     product_units.append(np.sum(u))
    
product_revenues = product_units * selling_price

for p, r in zip(products,product_revenues):
    print(f"product_revenue of {p} : {r}")
```

```python
# Q3. Har product ka total profit nikalo.

product_profits = (selling_price - cost_price) * product_units
for prd,prf in zip(products,product_profits):
    print(f"Profit of {prd} : {prf}")
```

```python
# Q4. Har product ka profit margin percentage calculate karo: total profit ÷ total revenue × 100. 

pmp = product_profits / product_revenues * 100        #  Profit Margin Percentage

for prd,pm in zip(products,pmp):
    print(f"PMP of {prd} : {np.round(pm)}%")
```

```python
                                Part B — Business Performance
                                    Aggregation & ranking
```

```python
# Q5. Har month ka total revenue nikalo.
# revenues = []
# for us,sp in zip(units_sold,selling_price):
#     rev_row = []
#     for u in us:
#         rev_row.append(u*sp)
#     revenues.append(rev_row)
# month_revenues = np.sum(revenues,axis = 0)

revenues = units_sold * selling_price[:, np.newaxis]
month_revenues = np.sum(revenues, axis = 0)

for m,mr in zip(months,month_revenues):
    print(f"Revenue of {m} : {mr}")
```

```python
# Q6.Har month ka total profit nikalo.

costs = []
for us,cp in zip(units_sold,cost_price):
    cost_row = []
    for u in us:
        cost_row.append(u*cp)
    costs.append(cost_row)

month_costs = np.sum(costs,axis = 0)
month_profits = month_revenues - month_costs

# print(month_revenues)
# print(month_costs)
# print(month_profits)

for m,mp in zip(months,month_profits):
    print(f"Profit of {m} : {mp}")
```

```python
# Q7.Sabse zyada revenue aur sabse zyada profit dene wale product ka naam, value ke saath nikalo.

hri = np.argmax(product_revenues)          # Highest Revenue Index (hri)
hpi = np.argmax(product_profits)          # Highest Profit Index  (hpi)

print(f"Highest Revenue Product : {products[hri]} - {product_revenues[hri]}")
print(f"Highest Profit Product : {products[hpi]} - {product_profits[hpi]}")
```

```python
# Q8. Revenue ke basis par products ko highest se lowest order mein rank karo. Product names aur revenue saath mein dikhao.

rank = np.argsort(product_revenues)[::-1]    

for i in rank:
    print(f"{products[i]} - {product_revenues[i]}")
```

```python
                        Part C — Advanced Filtering
                            Multiple conditions
```

```python
# Q9.Un products ko identify karo jinka total revenue ₹20,00,000 se zyada hai aur profit margin 15% se kam hai.

indxx = (product_revenues > 2000000) & (pmp < 15)
print(f"{products[indxx]} - {product_revenues[indxx]} - {np.round(pmp[indxx],2)}%")
```

```python
# Q10. Har product ka monthly average units sold nikalo, phir un products ke naam nikalo jinka average 80 se zyada hai.

ams = np.mean(units_sold, axis = 1)         # Average Monthly Sale
print(products[ams > 80])
```

```python
# Q11. Har product ka highest-sales month identify karo. Product name, month aur units sold display karo.

hms_indx = np.argmax(units_sold, axis = 1)             # Highest Monthly Sale Index
hms = np.max(units_sold, axis = 1)

# print(hms_indx)
# print(hms)
for p,hi,h in zip(products,hms_indx,hms):
    print(f"{p} - {months[hi]} - {h}")
```

```python
# Q12.Har month mein sabse zyada units sell karne wala product identify karo.

hps_indx = np.argmax(units_sold, axis = 0)               # Highest Sold Product in Months

for hi,m in zip(hps_indx,months):
    print(f"{m} - {products[hi]}")
```

```python
                                   Part D — Trend & Insights
                                        Hard challenge
```

```python
# Q13. April aur January revenue ka absolute difference aur percentage change calculate karo, product-wise. 

jan_indx = months == 'Jan'                  # Jan Index
apr_indx = months == 'Apr'                  # Apr Index

# sp_size = np.size(selling_price)
rev = []
for sp,i in zip(selling_price,range(np.size(selling_price))):
    rev_row = []
    for us in units_sold[i]:
        rev_row.append(us * sp)
    rev.append(rev_row)
    
rev = np.array(rev)

# print(rev)
rev_diff = rev[:,apr_indx] - rev[:,jan_indx]                   # Revenue Difference
chng_per = np.round(100*rev_diff/rev[:,jan_indx],2)            # Change Percentage

for p,rd,cp in zip(products,rev_diff,chng_per):
    print(f"{p} - {rd} - {cp}%")
```

```python
# Q14. Har product ke liye March se April tak units sold badhe, ghate ya same rahe—classify karo. 

mar_indx = months == 'Mar'
sale_diffma = units_sold[:,apr_indx] - units_sold[:,mar_indx]           #  Sale Diffrence of March And April

sale_stat = np.select([sale_diffma > 0 , sale_diffma < 0] , ['Sale Increase','Sale Decrease'], default = 'Sale as it')

for p,ss in zip(products,sale_stat):
    print(f"{p} - {ss}")
```

```python
# Q15. Har month ka profit margin calculate karo: total monthly profit ÷ total monthly revenue × 100.

spu = units_sold * selling_price[:,np.newaxis]            # Selling Prices for all units
cpu = units_sold * cost_price[:,np.newaxis]               # Cost Prices for all units

pu = spu - cpu                                           # Prrofits for all units
mp = np.sum(pu, axis = 0)                               # Monthly Profit 
mr = np.sum(rev, axis = 0)                              # Monthly Revenue

mpm = np.round((mp / mr) * 100,2)                       # Months Profit Margin

for m,mp in zip(months,mpm):
    print(f"{m} - {mp}%")
```
